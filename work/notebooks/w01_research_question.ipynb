## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

### Research question

Can observable search and content signals help prioritize which pages should be reviewed first for a possible content refresh?

**Decision:** Which pages should a content reviewer look at first?

**Action:** A reviewer could inspect the highest-ranked pages and decide whether they need updating, improvement, or no action.

**Cost of a wrong recommendation:** If a page is ranked too high, a reviewer may waste time reviewing a page that does not need attention. If an important page is ranked too low, the team may miss an opportunity to review it. The goal is therefore to improve prioritization, not to guarantee that a refresh will improve performance.

```python
import os, sys, subprocess

IN_COLAB = "google.colab" in sys.modules
REPO_URL = "https://github.com/flyrank-bih/flyrank-ml-internship-starter"
REPO_DIR = "flyrank-ml-internship-starter"

if IN_COLAB:
    if not os.path.isdir(REPO_DIR):
        subprocess.run(["git", "clone", "--depth", "1", REPO_URL, REPO_DIR], check=True)
    os.chdir(REPO_DIR)
    subprocess.run([
        sys.executable, "-m", "pip", "install", "-q", "-r", "requirements.txt"
    ], check=True)
elif os.path.basename(os.getcwd()) == "notebooks":
    os.chdir("..")

import pandas as pd
import numpy as np

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# A page is declining when its recent trend is down.
df["is_declining_label"] = (
    df["trend_direction"].str.lower().eq("down").astype(int)
)

# Search volume vs impressions
corr = df["search_volume"].corr(df["impressions_90d"])

print(df.shape[0], "pages | declining rate:",
      round(df["is_declining_label"].mean(), 3))

print("Search volume vs impressions correlation:",
      round(corr, 3))

# Pages that are stale and still visible
stale = df["days_since_last_update"] >= 180
visible = df["impressions_90d"] >= 500

print("Pages stale for 180+ days:", stale.sum())
print("Pages both stale and visible:", (stale & visible).sum())
```

The starter dataset contains 30,000 pages. I found that a number of pages have not been updated for 180 days or more, and some of these pages are still receiving substantial impressions. This makes the refresh lane worth investigating because there may be pages with enough visibility to justify human review even though they have become stale.
